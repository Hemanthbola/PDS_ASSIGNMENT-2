TASK A

```python
# Importing necessary libraries
import pandas as pd

# Load the dataset
df = pd.read_csv('train.csv')
```

```python
# Displaying the first few rows of the dataset
print("First 5 rows of the dataset:")
print(df.head())
```

```python
# Checking for missing values in each column
print("\nMissing values in each column:")
missing_values = df.isnull().sum()
print(missing_values)
```

```python
# Define numerical and categorical columns:
# This automatically categorizes columns based on their data type.
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = df.select_dtypes(include=['object']).columns
```

```python
# Imputing missing values for numerical columns using the median
for col in num_cols:
    if df[col].isnull().sum() > 0:
        median_value = df[col].median()
        df[col] = df[col].fillna(median_value)
        print(f"Imputed missing values in numerical column '{col}' with median: {median_value}")
```

```python
# Imputing missing values for categorical columns using the mode
for col in cat_cols:
    if df[col].isnull().sum() > 0:
        mode_value = df[col].mode()[0]
        df[col] = df[col].fillna(mode_value)
        print(f"Imputed missing values in categorical column '{col}' with mode: {mode_value}")
```

```python
# Checking again for missing values to ensure all have been handled
print("\nMissing values after imputation:")
print(df.isnull().sum())
```

TASK B

```python
# Printing the first few rows to inspect the data before cleaning
print("Before cleaning:")
print(df.head())
```

```python
# Defining a mapping for the columns and their corresponding unit strings.
# Adjusting column names if they differ in the dataset.
unit_removals = {
    'Mileage': 'kmpl',
    'Engine': 'CC',       # Example: values like "1200 CC"
    'Power': 'bhp',       # Example: values like "100 bhp"
    'New_price': 'lakh'   # Example: values like "5 lakh"
}
```

```python
# Loop through each column and remove the specified unit string.
for col, unit in unit_removals.items():
    if col in df.columns:
        # Removing the unit substring and any surrounding whitespace.
        # The regex=False parameter is used since we are replacing a literal substring.
        df[col] = df[col].astype(str).str.replace(unit, '', regex=False).str.strip()
        # Converting the cleaned string to a numeric type.
        df[col] = pd.to_numeric(df[col], errors='coerce')
        print(f"Processed column '{col}': removed unit '{unit}'")
```

```python
# Printing the first few rows to inspect the data after cleaning
print("\nAfter cleaning:")
print(df.head())
```

```python
# Displaying the initial columns and a few rows of the dataset
print("Before One-Hot Encoding:")
print(df[['Fuel_Type', 'Transmission']].head())
```

```python
# Performing one-hot encoding for categorical variables "Fuel_Type" and "Transmission"
# The drop_first parameter can be set to True to avoid multicollinearity if needed,
# but here we keep all dummy variables to preserve full information.
df_encoded = pd.get_dummies(df, columns=['Fuel_Type', 'Transmission'], drop_first=False)
```

```python
# Displaying the new columns and a few rows to verify the transformation
print("\nAfter One-Hot Encoding:")
print(df_encoded.head())
```

TASK E

```python
#Verify that the 'Year' column exists and create 'Car_Age'
if 'Year' not in df.columns:
    raise KeyError("The column 'Year' is missing from the dataset.")
```

```python
# Calculate the current year and create the Car_Age column
current_year = datetime.datetime.now().year
df['Car_Age'] = current_year - df['Year']
```

```python
selected_columns = ["Car_Age", "Price", "Mileage", "Fuel_Type", "Transmission"]
df_selected = df[selected_columns]
print("Selected Columns (first 5 rows):")
print(df_selected.head())
```

```python
# Filter the DataFrame
# Filter: Selecting cars that are less than 10 years old and have a Price greater than 2
df_filtered = df_selected[(df_selected["Car_Age"] < 10) & (df_selected["Price"] > 2)]
print("\nFiltered Data (first 5 rows):")
print(df_filtered.head())
```

```python
# Rename - Change "Price" to "Cost"
if 'Price' in df_filtered.columns:
    df_renamed = df_filtered.rename(columns={"Price": "Cost"})
    print("\nRenamed 'Price' to 'Cost'. Updated columns:")
    print(df_renamed.columns.tolist())
    print("First 5 rows after renaming:")
    print(df_renamed.head())
else:
    df_renamed = df_filtered.copy()
    print("\nColumn 'Price' not found; no renaming applied.")
```

```python
# Mutate - Create a new feature "Price_per_Year"
# To handle potential division by zero for new cars (Car_Age == 0), we set Price_per_Year equal to Cost when Car_Age is 0.
df_renamed['Price_per_Year'] = df_renamed.apply(
    lambda row: row['Cost'] if row['Car_Age'] == 0 else row['Cost'] / row['Car_Age'],
    axis=1
)
print("\nData after adding 'Price_per_Year':")
print(df_renamed.head())
```

```python
# Arrange - Sorting the DataFrame by "Cost" in descending order
df_arranged = df_renamed.sort_values(by="Cost", ascending=False)
print("\nData arranged by 'Cost' (descending order, first 5 rows):")
print(df_arranged.head())
```

```python
# Summarize with Group By: Calculating average cost and count of cars by Fuel_Type

summary = df_arranged.groupby("Fuel_Type").agg(
    avg_cost=("Cost", "mean"),
    count=("Cost", "count")
).reset_index()

print("\nSummary statistics grouped by 'Fuel_Type':")
print(summary)
```
